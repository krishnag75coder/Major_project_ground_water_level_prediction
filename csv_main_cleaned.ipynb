```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('CGWB_data_wide.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# Step 1: Drop columns (dates) with more than 85% missing values
df = df.dropna(thresh=df.shape[0] * 0.15, axis=1)

# Step 2: Identify date columns (from column index 7 onward in your dataset)
date_cols = df.columns[7:]

# Step 3: Fill missing values row-wise using median
df[date_cols] = df[date_cols].T.fillna(df[date_cols].median(axis=1)).T

# Final check
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
df.drop(['Unnamed: 0'],axis=1,inplace = True)
```
