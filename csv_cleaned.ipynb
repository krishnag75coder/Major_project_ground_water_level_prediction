```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('CGWB_data_wide.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# Step 1: Drop columns (dates) with more than 85% missing values
df = df.dropna(thresh=df.shape[0]*0.15, axis=1)

# Step 2: Identify date columns (from column index 7 onward in your dataset)
date_cols = df.columns[7:]

# Step 3: Forward + Backward fill (good for short gaps)
df[date_cols] = df[date_cols].ffill(axis=1).bfill(axis=1)

# Step 4: Interpolation across dates (linear filling between known points)
df[date_cols] = df[date_cols].interpolate(method="linear", axis=1)

# Step 5: Fill any remaining NaN values with the row’s median (seasonal fallback)
df[date_cols] = df[date_cols].T.fillna(df[date_cols].median(axis=1)).T

# Check how many null values are left
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
df.drop(['Unnamed: 0'],axis=1,inplace = True)
```
